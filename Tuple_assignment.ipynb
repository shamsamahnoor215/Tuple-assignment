Question 7: Looping & Filtering Values  
Given a tuple of test scores:  
scores = (45, 78, 88, 30, 92, 60, 55)  


Write a for loop that iterates through scores.  
If a score is 70 or above, print "<score> - Pass".
Otherwise, print "<score> - Fail".


```python
scores = (45, 78, 88, 30, 92, 60, 55)
for scores in scores:
   if scores>= 70:
     print(f"{scores}: pass")
   else:
     print(f"{scores}:fail")
```

Question 8: While Loop Traversal  
Write a while loop that prints each element in the tuple below along with its index, starting from index 0:  
cities = ("Lahore", "Karachi", "Islamabad", "Peshawar")  


Expected Output Format:
City 0: Lahore
City 1: Karachi
...




```python
cities = ("Lahore", "Karachi", "Islamabad", "Peshawar")

index = 0

while index < len(cities):
    print(f"City {index}: {cities[index]}")
    index += 1
```
